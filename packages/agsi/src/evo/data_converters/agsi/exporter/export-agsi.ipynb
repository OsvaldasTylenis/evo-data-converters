# Export Evo geological sections to an AGSi file

This notebook shows how to download Evo `geological-sections` objects and export them to an AGSi 1.0.1 package (`.agsi` ZIP file).

In this first cell we create a ServiceManagerWidget which will open a browser window and ask you to login.

Once logged in, a widget will be displayed below allowing you to select an organisation and workspace to export objects from.

__Required:__ You must fill in your `client_id` before running this cell.

```python
from evo.notebooks import ServiceManagerWidget

manager = await ServiceManagerWidget.with_auth_code(client_id="your-client-id").login()
```

In the cell below we specify the Evo Object UUID of each object we want to export and the output file path, and then call `export_agsi()`.

You may also specify the version of each object to export. If not specified, the latest version is exported.

__Required:__ You must fill in your `object_id` of the Evo Object to export before running this cell.

_Note:_ Only `geological-sections` objects are supported. Each section must be a vertical plane, and each volume must be a single ring that splits into one top and one bottom profile (no holes or overhangs); otherwise `export_agsi()` raises an error and no file is written. See the package README for the full mapping.

```python
import os
from uuid import UUID

from evo.data_converters.common import EvoObjectMetadata
from evo.data_converters.agsi.exporter import export_agsi

# List of object_id and version_id pairs
objects_to_export = [
    {"object_id": "", "version_id": ""},
    {"object_id": "", "version_id": None},
    # Add more objects as needed
]
objects = []
for object_to_export in objects_to_export:
    object_id = object_to_export["object_id"]
    version_id = object_to_export["version_id"]
    objects.append(EvoObjectMetadata(object_id=UUID(object_id), version_id=version_id))

output_dir = "data/output"
os.makedirs(output_dir, exist_ok=True)

output_file = f"{output_dir}/object.agsi"

export_agsi(
    filepath=output_file,
    objects=objects,
    service_manager_widget=manager,
)

print("Done")
```
